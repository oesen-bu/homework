# Section 1: Introduction

## Dataset: Bike Rentals (Option 3 — Outdoor Recreation)

This project analyzes hourly bike rental data to understand how weather conditions, seasons, and time patterns affect bike usage. The dataset contains hourly rental counts along with environmental variables like temperature, humidity, and wind speed.

**Questions to investigate:**
- How do weather and seasonal factors influence bike rental demand?
- Are there differences in rental patterns between working days and non-working days?
- Which weather variables are most strongly related to rental counts?
- Can we predict bike rental counts using weather and time features?

# Section 2: Load and Inspect the Data

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np

df = pd.read_csv("bikes_final.csv")
df.head(10)
```

```python
print("Shape:", df.shape)
print("\nColumn types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
```

# Section 3: Data Cleaning and Preparation

We convert the `datetime` column to a proper datetime type and extract useful time components. We also map numeric codes to readable labels for season and weather.

```python
# Convert datetime column
df["datetime"] = pd.to_datetime(df["datetime"])

# Extract time components
df["hour"] = df["datetime"].dt.hour
df["month"] = df["datetime"].dt.month
df["year"] = df["datetime"].dt.year

# Map season and weather codes to readable labels
season_map = {1: "Spring", 2: "Summer", 3: "Fall", 4: "Winter"}
weather_map = {1: "Clear", 2: "Mist/Cloudy", 3: "Light Rain/Snow", 4: "Heavy Rain/Snow"}

df["season_label"] = df["season"].map(season_map)
df["weather_label"] = df["weather"].map(weather_map)

df.head()
```

# Section 4: Descriptive Statistics

```python
# Descriptive statistics for numerical columns
numerical_cols = ["temp", "atemp", "humidity", "windspeed", "casual", "registered", "count"]
df[numerical_cols].describe()
```

```python
# Categorical column summaries
print("Season distribution:")
print(df["season_label"].value_counts())
print("\nWeather distribution:")
print(df["weather_label"].value_counts())
print("\nDay of week distribution:")
print(df["day_of_week"].value_counts())
print("\nWorking day distribution:")
print(df["workingday"].value_counts())
```

# Section 5: Numerical Data Visualizations

We use histograms and box plots to show the distribution of key numerical variables.

```python
# Helper function for creating labeled histograms (reused throughout)
def plot_histogram(data, title, xlabel, bins=30, color="steelblue"):
    plt.figure(figsize=(8, 4))
    plt.hist(data, bins=bins, color=color, edgecolor="black")
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

plot_histogram(df["count"], "Distribution of Bike Rental Counts", "Rental Count")
plot_histogram(df["temp"], "Distribution of Temperature", "Temperature (°C)")
plot_histogram(df["humidity"], "Distribution of Humidity", "Humidity (%)")
```

```python
# Box plots for numerical variables
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

cols_to_plot = ["count", "temp", "windspeed"]
titles = ["Rental Count", "Temperature (°C)", "Wind Speed"]

for ax, col, title in zip(axes, cols_to_plot, titles):
    ax.boxplot(df[col].dropna())
    ax.set_title(title)
    ax.set_ylabel(title)

plt.suptitle("Box Plots of Key Numerical Variables", fontsize=14)
plt.tight_layout()
plt.show()
```

The rental count distribution is right-skewed, meaning most hours have relatively low rentals with occasional peaks. Temperature follows a roughly normal distribution. Wind speed is right-skewed with some high outliers.

# Section 6: Categorical Data Analysis

We visualize the distribution of categorical variables using bar charts and a pie chart.

```python
# Bar chart: records per season
season_counts = df["season_label"].value_counts()
plt.figure(figsize=(8, 4))
season_counts.plot(kind="bar", color=["green", "orange", "brown", "skyblue"], edgecolor="black")
plt.title("Number of Records by Season")
plt.xlabel("Season")
plt.ylabel("Count of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Pie chart: weather condition distribution
weather_counts = df["weather_label"].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(weather_counts, labels=weather_counts.index, autopct="%1.1f%%", startangle=140)
plt.title("Weather Condition Distribution")
plt.tight_layout()
plt.show()
```

The majority of recorded hours have clear weather. Seasons are roughly evenly distributed. This makes sense since the data covers full calendar years.

# Section 7: Grouped Analysis

We group the data by season and day type to compare rental patterns.

```python
# Average rentals by season
season_group = df.groupby("season_label")["count"].mean().sort_values(ascending=False)
print("Average hourly rentals by season:")
print(season_group)

plt.figure(figsize=(8, 4))
season_group.plot(kind="bar", color="coral", edgecolor="black")
plt.title("Average Hourly Bike Rentals by Season")
plt.xlabel("Season")
plt.ylabel("Average Rentals")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Average rentals by working day vs non-working day
workday_group = df.groupby("workingday")[["casual", "registered", "count"]].mean()
workday_group.index = ["Non-Working Day", "Working Day"]
print("Average rentals by day type:")
print(workday_group)

workday_group[["casual", "registered"]].plot(kind="bar", figsize=(8, 4), edgecolor="black")
plt.title("Average Casual vs Registered Rentals by Day Type")
plt.xlabel("Day Type")
plt.ylabel("Average Rentals")
plt.xticks(rotation=0)
plt.legend(title="User Type")
plt.tight_layout()
plt.show()
```

Fall has the highest average rentals, while spring has the lowest. Casual riders are much more active on non-working days, while registered riders dominate on working days — suggesting commuters vs. recreational users.

# Section 8: New Feature Engineering

We create a **casual ratio** column — the percentage of total rentals that come from casual (non-registered) users. This helps us distinguish hours driven by recreational riders from commuter-heavy hours.

```python
# New column: casual ratio (% of rentals that are casual users)
df["casual_ratio"] = df["casual"] / df["count"] * 100

# Show how casual ratio varies by hour of day
hourly_casual = df.groupby("hour")["casual_ratio"].mean()

plt.figure(figsize=(10, 4))
plt.bar(hourly_casual.index, hourly_casual.values, color="mediumpurple", edgecolor="black")
plt.title("Average Casual Rider Ratio by Hour of Day")
plt.xlabel("Hour")
plt.ylabel("Casual Ratio (%)")
plt.xticks(range(0, 24))
plt.tight_layout()
plt.show()
```

The casual ratio is highest during midday and overnight hours, and lowest during the 7-9 AM and 5-6 PM rush hours. This confirms that registered users are mostly commuters while casual users ride recreationally.

# Section 9: Trends Over Time

We look at monthly rental trends over the two years in the dataset, using standardized values so we can compare rental count and temperature on the same scale.

```python
# Monthly averages over time
df["year_month"] = df["datetime"].dt.to_period("M")
monthly = df.groupby("year_month")[["count", "temp"]].mean()
monthly.index = monthly.index.astype(str)

# Raw trends
plt.figure(figsize=(12, 5))
plt.plot(monthly.index, monthly["count"], marker="o", label="Avg Rentals")
plt.title("Monthly Average Bike Rentals Over Time")
plt.xlabel("Month")
plt.ylabel("Average Hourly Rentals")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Standardized comparison: rentals vs temperature
def standardize(series):
    return (series - series.mean()) / series.std()

plt.figure(figsize=(12, 5))
plt.plot(monthly.index, standardize(monthly["count"]), marker="o", label="Rentals (standardized)")
plt.plot(monthly.index, standardize(monthly["temp"]), marker="s", label="Temperature (standardized)")
plt.title("Standardized Monthly Trends: Rentals vs Temperature")
plt.xlabel("Month")
plt.ylabel("Standardized Value (z-score)")
plt.xticks(rotation=45)
plt.legend()
plt.axhline(y=0, color="gray", linestyle="--", linewidth=0.8)
plt.tight_layout()
plt.show()
```

Bike rentals follow a strong seasonal pattern — peaking in summer and dropping in winter. The standardized plot shows that rental count closely tracks temperature throughout the year, suggesting temperature is a major driver of demand. Year 2 also shows generally higher rentals, suggesting growing popularity.

# Section 10: Variable Relationships

We compute correlations among numerical variables and use scatter plots to explore relationships.

```python
# Correlation matrix
corr_cols = ["temp", "atemp", "humidity", "windspeed", "casual", "registered", "count"]
corr_matrix = df[corr_cols].corr()
print("Correlation matrix:")
print(corr_matrix.round(3))
```

```python
# Scatter plots: temp vs count, humidity vs count
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(df["temp"], df["count"], alpha=0.2, s=10, color="tomato")
axes[0].set_title("Temperature vs Rental Count")
axes[0].set_xlabel("Temperature (°C)")
axes[0].set_ylabel("Rental Count")

axes[1].scatter(df["humidity"], df["count"], alpha=0.2, s=10, color="teal")
axes[1].set_title("Humidity vs Rental Count")
axes[1].set_xlabel("Humidity (%)")
axes[1].set_ylabel("Rental Count")

plt.tight_layout()
plt.show()
```

Temperature has a moderate positive correlation with rental count — warmer weather drives more ridership. Humidity has a weak negative correlation — very humid conditions discourage biking slightly.

# Section 11: Hypothesis

**Hypothesis:** Higher temperatures lead to higher bike rental counts. People are more likely to rent bikes when the weather is warm and pleasant. We expect a positive linear relationship between temperature and hourly rental count.

# Section 12: Linear Regression

We use `sklearn`'s `LinearRegression` to model the relationship between temperature and rental count. We also build a multi-feature model to see if adding humidity and wind speed improves the prediction.

```python
# Simple linear regression: temp -> count
X_simple = df[["temp"]]
y = df["count"]

model_simple = LinearRegression()
model_simple.fit(X_simple, y)
y_pred_simple = model_simple.predict(X_simple)

print(f"Simple model (temp only):")
print(f"  Coefficient: {model_simple.coef_[0]:.3f}")
print(f"  Intercept: {model_simple.intercept_:.3f}")
print(f"  R² score: {model_simple.score(X_simple, y):.4f}")

# Plot actual vs predicted with regression line
plt.figure(figsize=(10, 5))
plt.scatter(df["temp"], y, alpha=0.15, s=10, color="steelblue", label="Actual")
# Sort by temp for a clean regression line
sort_idx = df["temp"].argsort()
plt.plot(df["temp"].iloc[sort_idx], y_pred_simple[sort_idx], color="red", linewidth=2, label="Regression Line")
plt.title("Linear Regression: Temperature vs Rental Count")
plt.xlabel("Temperature (°C)")
plt.ylabel("Rental Count")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Multi-feature regression: temp + humidity + windspeed -> count
X_multi = df[["temp", "humidity", "windspeed"]]
model_multi = LinearRegression()
model_multi.fit(X_multi, y)
y_pred_multi = model_multi.predict(X_multi)

print(f"Multi-feature model (temp + humidity + windspeed):")
print(f"  Coefficients: {dict(zip(X_multi.columns, model_multi.coef_.round(3)))}")
print(f"  Intercept: {model_multi.intercept_:.3f}")
print(f"  R² score: {model_multi.score(X_multi, y):.4f}")

# Predicted vs actual plot
plt.figure(figsize=(8, 6))
plt.scatter(y, y_pred_multi, alpha=0.1, s=10, color="darkorange")
plt.plot([0, y.max()], [0, y.max()], color="black", linestyle="--", linewidth=1, label="Perfect prediction")
plt.title("Predicted vs Actual Rental Count (Multi-Feature Model)")
plt.xlabel("Actual Count")
plt.ylabel("Predicted Count")
plt.legend()
plt.tight_layout()
plt.show()
```

## Regression Discussion

The simple model using only temperature gives a low R² — temperature alone explains only a small portion of the variance in hourly rental counts. Adding humidity and wind speed in the multi-feature model improves R² slightly, but the model still has clear limitations.

**What the results suggest:** Temperature has a positive effect on rentals, humidity has a negative effect, and wind speed has a small negative effect. The direction of these relationships matches our intuition.

**Limitations:** Weather variables alone cannot fully predict hourly bike rentals. Major factors like hour of day, whether it's a working day, and overall year-over-year growth are not included in this model. A more complete model would incorporate time-based features alongside weather.
